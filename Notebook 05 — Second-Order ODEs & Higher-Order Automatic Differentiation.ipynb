## Notebook 05 — Second-Order ODEs

### Problem

We want to solve:

    d²y/dt² + y = 0

with initial conditions:

    y(0) = 1
    dy/dt(0) = 0

The analytical solution is:

    y(t) = cos(t)

However, the PINN will NOT be given this solution during training.

Instead, it will learn a function:

    y_hat(t)

and will be required to satisfy:

    d²y_hat/dt² + y_hat = 0

as well as:

    y_hat(0) = 1

    dy_hat/dt(0) = 0

# 1. Simple Harmonic Motion

Consider an ideal spring.

Newton's second law gives:

    F = ma

For a spring:

    F = -kx

Therefore:

    m d²x/dt² = -kx

Rearranging:

    d²x/dt² + (k/m)x = 0

If we choose units where:

    k/m = 1

we get:

    d²x/dt² + x = 0

This is our differential equation.

It describes oscillatory motion.

The exact solution for:

    x(0) = 1
    x'(0) = 0

is:

    x(t) = cos(t)

```python
import numpy as np
import matplotlib.pyplot as plt

import torch
import torch.nn as nn

torch.manual_seed(42)
np.random.seed(42)
```

```python
t_exact = np.linspace(0, 20, 1000)

y_exact = np.cos(t_exact)

plt.figure(figsize=(10, 5))

plt.plot(t_exact, y_exact)

plt.xlabel("Time")
plt.ylabel("y(t)")
plt.title("Simple Harmonic Oscillation")
plt.grid(True)

plt.show()
```
